- **档案**：超过 50 年的连续图像

Landsat 无与伦比的时间深度使其对于城市扩张、森林砍伐或冰川消退等长期变化分析至关重要。

## 使用 STAC 进行搜索

### 什么是 STAC？

[时空资产目录 (STAC)](https://stacspec.org) 规范定义了一种通用的、基于 JSON 的语言来描述地理空间资产。在 STAC 之前，每个提供商都有自己的搜索界面和元数据模式，这使得跨提供商的搜索不切实际。

如今，主要平台都已采用 STAC，包括 Microsoft Planetary Computer、NASA Earthdata 和 Element 84 Earth Search。学习 STAC 可让您掌握一套适用于所有这些来源的技能。

### STAC 规范

STAC 规范将地理空间数据组织成四个层次组件：

- **目录**：链接到集合的顶级入口点。例如，Planetary Computer 目录包含 Microsoft 托管的所有地球观测数据集。

- **集合**：一组共享通用元数据（例如 [sentinel-2-l2a](https://planetarycomputer.microsoft.com/dataset/sentinel-2-l2a) 或 [naip](https://planetarycomputer.microsoft.com/dataset/naip)）的相关项目。集合描述数据集的空间/时间范围、许可证和可过滤属性。

- **项目**：代表单个时空观测的 GeoJSON 特征。每个项目包含几何、日期时间、属性（云量、平台等）以及其资产的链接。

- **资产**：与项目关联的单个文件，例如光谱波段文件 (COG)、缩略图或元数据文档。每个资产都有一个键、href、媒体类型和可选角色。

工作流程遵循此层次结构：连接到**目录**，浏览**集合**，使用空间/时间/属性过滤器搜索**项目**，然后访问特定的**资产**。

### STAC API 端点

**静态 STAC** 目录是 Web 服务器上的一组链接 JSON 文件。**STAC API** 通过 RESTful 搜索端点扩展了此功能：

- `GET /` -- 带有链接和 API 功能的着陆页
- `GET /collections` -- 列出所有可用的集合
- `GET /collections/{id}` -- 特定集合的元数据
- `POST /search` -- 使用过滤器跨集合进行搜索

搜索端点接受 `bbox`、`datetime`、`collections`、`query` 和 `max_items` 参数。实际上，`pystac_client` 库将这些抽象为 Pythonic 接口。

### STAC 的 Python 库

两个库构成了使用 STAC 的基础：

- [**pystac**](https://pystac.readthedocs.io/) 是用于在 Python 中读取、创建和操作 STAC 对象的核心库。

- [**pystac_client**](https://pystac-client.readthedocs.io/) 增加了 `Client` 类，用于连接到 STAC API 并执行搜索。

```text
project/
├── data/

└── results/ # 预测、地图和评估输出
```

遵循以下命名约定：

- **包含来源**：`naip_2021_knoxville.tif`、`s2_20240615_B04.tif`
- **包含日期**：使用 ISO 格式（YYYY-MM-DD 或 YYYYMMDD），以便文件按时间顺序排序
- **包含区域**：添加简短的位置标识符，例如城市名称或瓦片 ID
- **保留原始数据**：切勿修改原始下载。在单独的目录中创建处理后的副本。
- **使用一致的 CRS**：在工作流程的早期将所有数据集重新投影到公共 CRS，以避免后期错位

工作流程遵循此层次结构：连接到**目录**，浏览**集合**，使用空间/时间/属性过滤器搜索**项目**，然后访问特定的**资产**。

### STAC API 端点

**静态 STAC** 目录是 Web 服务器上的一组链接 JSON 文件。**STAC API** 通过 RESTful 搜索端点扩展了此功能：

- `GET /` -- 带有链接和 API 功能的着陆页
- `GET /collections` -- 列出所有可用的集合
- `GET /collections/{id}` -- 特定集合的元数据
- `POST /search` -- 使用过滤器跨集合进行搜索

搜索端点接受 `bbox`、`datetime`、`collections`、`query` 和 `max_items` 参数。实际上，`pystac_client` 库将这些抽象为 Pythonic 接口。

### STAC 的 Python 库

两个库构成了使用 STAC 的基础：

- [**pystac**](https://pystac.readthedocs.io/) 是用于在 Python 中读取、创建和操作 STAC 对象的核心库。

- [**pystac_client**](https://pystac-client.readthedocs.io/) 增加了 `Client` 类，用于连接到 STAC API 并执行搜索。

连接到 Planetary Computer 的 STAC API 并探索其集合：

## 下载 Sentinel-2 数据

工作流程分两步：使用 `pystac_client` 搜索项目，然后使用 `geoai.download_pc_stac_item()` 下载特定波段。

### 搜索 Sentinel-2 项目

````python
from pystac_client import Client

catalog = Client.open("https://planetarycomputer.microsoft.com/api/stac/v1")
bbox = [-83.94, 35.96, -83.92, 35.98]

search = catalog.search(
    collections=["sentinel-2-l2a"],
    bbox=bbox,
    datetime="2025-06-01/2025-08-31",
    query={"eo:cloud_cover": {"lt": 10}},
    max_items=1,
)

items = search.item_collection()
if items:
    item = items[0]
    cloud_cover = item.properties.get("eo:cloud_cover")
    cloud_cover_text = f"{cloud_cover}%" if cloud_cover is not None else "N/A"
    print(f"Selected item: {item.id}")
    print(f"Date: {item.datetime}")
    print(f"Cloud cover: {cloud_cover_text}")
    item_url = item.self_href
    print(f"Item URL: {item_url}")
````

```text
选定项目：S2B_MSIL2A_20250816T161829_R040_T16SGE_20250816T201128

项目 URL：https://planetarycomputer.microsoft.com/api/stac/v1/collections/sentinel-2-l2a/items/S2B_MSIL2A_20250816T161829_R040_T16SGE_20250816T201128
```

### 下载和合并波段

使用 `geoai.download_pc_stac_item()` 下载特定波段并可选择将其合并到单个多波段 GeoTIFF 中：

````python
if items:
    result = geoai.download_pc_stac_item(
        item_url=item_url,
        bands=["B02", "B03", "B04", "B08"],  # Blue, Green, Red, NIR
        output_dir="sentinel2_data",
        merge_bands=True,
        merged_filename="knoxville_s2_rgbn.tif",
        overwrite=False,
    )

    for band_name, path in result.items():
        print(f"  {band_name}: {path}")
````

`merge_bands=True` 参数将所有请求的波段重采样到公共分辨率，并将它们堆叠到单个 GeoTIFF 中。您可以使用 `cell_size` 参数控制输出分辨率。

### Overture Maps 建筑物

[Overture Maps 基金会](https://overturemaps.org) 生成开放地图数据，包括全球建筑物足迹数据集。`geoai` 包提供了两种访问此数据的方式。

`download_overture_buildings()` 函数下载边界框的建筑物数据：

````python
import geoai

bbox = (-83.94, 35.96, -83.92, 35.98)  # Knoxville area
output_path = "buildings.geojson"

geoai.download_overture_buildings(
    bbox=bbox,
    output=output_path,
    overture_type="building",
)
print(f"Buildings saved to {output_path}")
````

Overture Maps 函数需要 `overturemaps` 包（`pip install overturemaps`）。

### OpenStreetMap 数据

[OpenStreetMap (OSM)](https://www.openstreetmap.org) 是一个由志愿者维护的协作式开源地图。`osmnx` 和 `quackosm` 库（它使用 DuckDB 进行快速 PBF 提取）等库提供了查询 OSM 数据的 Python 接口。`leafmap` 包封装了这些工具，方便访问。

按边界框下载 OSM 建筑物足迹：

````python
import leafmap.osm as osm

bbox = (-83.94, 35.96, -83.92, 35.98)  # Knoxville area
buildings = osm.quackosm_gdf_from_bbox(bbox, tags={"building": True})
print(f"Downloaded {len(buildings)} buildings")
buildings.head()
````

按地名查询：

````python
roads = osm.quackosm_gdf_from_place("Knoxville, Tennessee", tags={"highway": True})
print(f"Downloaded {len(roads)} road segments")
roads.head()
````

按自定义几何体查询：

````python
from shapely.geometry import Polygon

polygon = Polygon(
    [
        (-83.94, 35.96),
        (-83.92, 35.96),
        (-83.92, 35.98),
        (-83.94, 35.98),
    ]
)
natural = osm.quackosm_gdf_from_geometry(polygon, tags={"natural": True})
print(f"Downloaded {len(natural)} natural features")
natural.head()
````

GeoAI 中常用的 OSM 标签包括 `building`、`highway`、`landuse`、`natural`、`waterway` 和 `amenity`。某个区域的首次下载可能会比较慢，因为 PBF 文件需要缓存；后续查询会重用它。

## 组织您的数据

清晰的目录结构可以防止混淆并使工作流程可重现。建议的布局：

```text
project/
├── data/
│   ├── raw/ # 原始下载文件（从不修改）
│   │ ├── naip/
│   │ ├── sentinel2/
│   │ ├── landsat/
│   │ └── vectors/
│   ├── processed/ # 裁剪、重新投影或合并的文件
│   └── training/ # 准备好用于模型训练的瓦片和标签
├── models/ # 保存的模型权重和配置
├── notebooks/ # 用于探索的 Jupyter 笔记本
├── scripts/ # 可重用的 Python 脚本
└── results/ # 预测、地图和评估输出
```

遵循以下命名约定：

- **包含来源**：`naip_2021_knoxville.tif`、`s2_20240615_B04.tif`
- **包含日期**：使用 ISO 格式（YYYY-MM-DD 或 YYYYMMDD），以便文件按时间顺序排序
- **包含区域**：添加简短的位置标识符，例如城市名称或瓦片 ID
- **保留原始数据**：切勿修改原始下载。在单独的目录中创建处理后的副本。
- **使用一致的 CRS**：在工作流程的早期将所有数据集重新投影到公共 CRS，以避免后期错位
